# Setting up WASIM for a Swiss basin on Euler

- connect to the VPN and Euler

## model build

- fork the WASIM github repo and clone it to your `$HOME/` directory
https://github.com/jschulla-wasimeth/wasim-eth

- clone WASIM_tutorial repo to your `$HOME/` directory
https://github.com/aelkouk/WASIM_tutorial

- load modules
`module load stack/2024-06 gcc/12.2.0 openmpi/4.1.6 netcdf-c/4.9.2`

- save environement
`module save env-wasim11`

- copy makefile
`cp ~/WASIM_tutorial/makefile_euler_mpi ~/wasim-eth/src/`

- build model `make -f makefile_euler_mpi`

- excute
`./wasim_mpi_11-00-10-x86-64-pc-linux-euler_intel`

## basin setup 

- copy swiss basins setups
`cp /cluster/home/aelkouk/WASIM_setups.zip $HOME/`

- unzip
`unzip WASIM_setups.zip -d WASIM_setups/`

- make a project in your `$HOME`:
`mkdir $HOME/WASIM_retreat/`

- choose a basin to work on
`cp -r $HOME/WASIM_setups/Rhone/ $HOME/WASIM_retreat/`

- Download the basin folder to your laptop (FileZilla)

Inputs inspection:
- Open Control File: which features are activate and which methods are used (e.g., interpolation of precipitation, soils, ET)
- Load the input raster grids (on QGIS): dem, zones, subcatchments, river channel
- Load precipitation stations (on QGIS)
- Timeseries: plot precipitation and temperatures over a year for a station of your choice
